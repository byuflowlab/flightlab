```python
from flightlab import atmos, drag, geom
from flightlab.fleet import C172

V = 68
h = 2200
air = atmos.at(h)
panel = geom.resolve(C172.wing)
buildup = drag.buildup(
    C172, V, altitude=h,
    interference=0.05,
    protuberance=0.05,
)
print(air.density, air.q(V), air.mach(V))
print(air.reynolds(V, panel.mac))
print(buildup.table())# Paste generated Python here, then extend it into your own analysis.
```
